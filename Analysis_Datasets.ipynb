# Shoppers dataset analysis

Contains a distribution-level analysis of both numerical and categorical columns, as well as
correlation matrices, missing values, class imbalance, and outliers

```python
df_shoppers = pd.read_csv('data/shoppers/dataset.csv')

# 3. Initial view
# 3.1. Show the first rows
print("Primeras 5 filas:")
display(df_shoppers.head())

# 3.2. General information about columns, types, and non-null values
print("\nInformación del DataFrame:")
display(df_shoppers.info())

# 3.3. Descriptive statistics for numerical variables
print("\nEstadísticos descriptivos (numéricos):")
display(df_shoppers.describe().T)

# 3.4. Descriptive statistics for categorical variables
print("\nEstadísticos descriptivos (categóricos):")
display(df_shoppers.describe(include=['object', 'category']).T)

# 4. Missing values detection
print("\nValores faltantes por columna:")
display(df_shoppers.isnull().sum())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure theme and context for paper publications
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)

# Load the dataset
df_shoppers = pd.read_csv('data/shoppers/dataset.csv')

# Identify numerical and categorical variables
num_cols = df_shoppers.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_shoppers.select_dtypes(include=['object', 'category']).columns.tolist()

# 1. Histograms + KDE for numerical variables
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df_shoppers[col].dropna(), kde=True, color="steelblue", edgecolor="black")
    plt.title(f"Distribution of {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frequency", fontsize=12)
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()
```

```python


# 2. Countplot for categorical variables
for col in cat_cols:
    plt.figure(figsize=(6, 4))
    order = df_shoppers[col].value_counts().index
    sns.countplot(data=df_shoppers, x=col, order=order, palette="pastel", edgecolor="black")
    plt.title(f"Count of {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frequency", fontsize=12)
    plt.xticks(rotation=45, ha='right')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

```

```python
# Identify numerical and categorical variables
num_cols = df_shoppers.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_shoppers.select_dtypes(include=['object', 'category']).columns.tolist()


# 3. Heatmap of correlations between numerical variables
plt.figure(figsize=(20, 8))
corr = df_shoppers[num_cols].corr()
sns.heatmap(corr, annot=True, fmt=".1f", cmap="vlag", cbar=True, square=True, linewidths=.5)
plt.title("Shopper heatmap", fontsize=16)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.savefig('results/exploratory_analysis/correlaciones_shopper.png')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure theme and context for paper publications
#sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)

# Identify numerical and categorical variables
num_cols = df_shoppers.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_shoppers.select_dtypes(include=['object', 'category']).columns.tolist()

# 1. Histograms + KDE for numerical variables
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df_shoppers[col].dropna(), kde=True, color="steelblue", edgecolor="black")
    plt.title(f"Distribution of {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frequency", fontsize=12)
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

# 2. Countplot for categorical variables
for col in cat_cols:
    plt.figure(figsize=(6, 4))
    order = df_shoppers[col].value_counts().index
    sns.countplot(data=df_shoppers, x=col, order=order, palette="pastel", edgecolor="black")
    plt.title(f"Count of {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frequency", fontsize=12)
    plt.xticks(rotation=45, ha='right')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

# 4. Boxplots to detect outliers in numerical variables
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df_shoppers[col], color="lightseagreen")
    plt.title(f"Boxplot of {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

# 5. Example of bivariate boxplot (numerical vs. categorical)
if len(cat_cols) > 0 and len(num_cols) > 0:
    ejemplo_cat = cat_cols[0]
    ejemplo_num = num_cols[0]
    plt.figure(figsize=(8, 5))
    sns.boxplot(x=ejemplo_cat, y=ejemplo_num, data=df_shoppers, palette="Set2")
    plt.title(f"{ejemplo_num} by {ejemplo_cat}", fontsize=16)
    plt.xlabel(ejemplo_cat, fontsize=12)
    plt.ylabel(ejemplo_num, fontsize=12)
    plt.xticks(rotation=45, ha='right')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Identify numerical variables
num_cols = df_shoppers.select_dtypes(include=['number']).columns.tolist()
n = len(num_cols)

# Define subplot layout: up to 3 columns per row, adjust rows
max_cols_per_row = 3
n_cols_subplot = min(n, max_cols_per_row)
n_rows_subplot = math.ceil(n / n_cols_subplot)

# Create figure with subplots
fig, axes = plt.subplots(
    n_rows_subplot,
    n_cols_subplot,
    figsize=(4 * n_cols_subplot, 4 * n_rows_subplot),
    sharey=False
)

# Flatten the axes matrix to iterate easily
axes_flat = axes.flatten() if n > 1 else [axes]

# Generate each boxplot in its corresponding axis
for idx, col in enumerate(num_cols):
    ax = axes_flat[idx]
    sns.boxplot(
        x=df_shoppers[col],
        ax=ax,
        color="lightseagreen"
    )
    ax.set_title(f"{col}", fontsize=12)
    ax.set_xlabel("")  # Remove x-axis label (name only appears in title)
    ax.set_ylabel("")  # Optional: if you prefer only numbers without label

# Deactivate extra subplots (if n is not a multiple of 3)
for j in range(n, len(axes_flat)):
    axes_flat[j].axis('off')

# --------------------------------------------------------------------------------
# Important: set the main title *before* calling tight_layout, 
# and tell tight_layout to reserve space for the suptitle via `rect`.
# --------------------------------------------------------------------------------
plt.suptitle("Numerical Variables Boxplots", fontsize=16)

# Adjust tight_layout leaving top space for the suptitle.
# The rect=[0, 0, 1, 0.95] indicates that the subplots area occupies from 
# 0% to 95% of the figure height (leaving 5% at the top for the title).
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Save figure making sure everything fits within the limits
plt.savefig('results/exploratory_analysis/boxplot_adult.png', bbox_inches='tight')

# Finally we show on screen
plt.show()
```

```python
df_default = pd.read_csv('data/default/dataset.csv')

# Identify numerical and categorical variables
num_cols = df_default.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_default.select_dtypes(include=['object', 'category']).columns.tolist()


# 3. Heatmap of correlations between numerical variables
plt.figure(figsize=(20, 10))
corr = df_default[num_cols].corr()
sns.heatmap(corr, annot=True, fmt=".1f", cmap="vlag", cbar=True, square=True, linewidths=.5)
plt.title("Default heatmap", fontsize=16)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.savefig('results/exploratory_analysis/correlaciones_default.png')
plt.show()
```

```python
# --------------------------------------------------
# 1. Load the three datasets and define their target
# --------------------------------------------------

# 1.1. Shoppers (already loaded before); target: 'income'
df_shoppers = pd.read_csv('data/shoppers/dataset.csv')
target_shoppers = 'Revenue'

# 1.2. Adult; target: 'income'
df_adult = pd.read_csv('data/adult/dataset.csv')
df_adult['income'] = df_adult['income'].str.replace('.', '', regex=False)
target_adult = 'income'

# 1.3. Default; target: 'default'
df_default = pd.read_csv('data/default/dataset.csv')
target_default = 'Y'

# --------------------------------------------------
# 2. Prepare figure and axes (1 row, 3 columns)
# --------------------------------------------------

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(18, 5),
    sharey=False
)

# --------------------------------------------------
# 3. Plot #1: Shoppers – Count of 'income'
# --------------------------------------------------

ax = axes[0]
order_sh = df_shoppers[target_shoppers].value_counts().index
sns.countplot(
    data=df_shoppers,
    x=target_shoppers,
    order=order_sh,
    palette="pastel",
    edgecolor="black",
    ax=ax
)

ax.set_title("Shoppers count", fontsize=14)
ax.set_xlabel("Revenue", fontsize=12)
ax.set_ylabel("Frequency", fontsize=12)
ax.tick_params(axis='x', rotation=45)
sns.despine(ax=ax, trim=True)

# --------------------------------------------------
# 4. Plot #2: Adult – Count of 'income'
# --------------------------------------------------

ax = axes[1]
order_ad = df_adult[target_adult].value_counts().index
sns.countplot(
    data=df_adult,
    x=target_adult,
    order=order_ad,
    palette="pastel",
    edgecolor="black",
    ax=ax
)

ax.set_title("Adult count", fontsize=14)
ax.set_xlabel("Income", fontsize=12)
ax.set_ylabel("")  # We no longer need to repeat the y-label
ax.tick_params(axis='x', rotation=45)
sns.despine(ax=ax, trim=True)

# --------------------------------------------------
# 5. Plot #3: Default – Count of 'default'
# --------------------------------------------------

ax = axes[2]
order_df = df_default[target_default].value_counts().index
sns.countplot(
    data=df_default,
    x=target_default,
    order=order_df,
    palette="pastel",
    edgecolor="black",
    ax=ax
)

ax.set_title("Default count", fontsize=14)
ax.set_xlabel("Y", fontsize=12)
ax.set_ylabel("")  # Same reasons as in the previous subplot
ax.tick_params(axis='x', rotation=45)
sns.despine(ax=ax, trim=True)

# --------------------------------------------------
# 6. Final adjustments and saving the figure
# --------------------------------------------------

plt.tight_layout(rect=[0, 0, 1, 0.95])  # leave space for a possible suptitle
# If you want a general title uncomment the following line:
# plt.suptitle("Distribution of target variables in three datasets", fontsize=16, y=1.02)
plt.savefig('results/exploratory_analysis/target_dis.png')
plt.show()
```
