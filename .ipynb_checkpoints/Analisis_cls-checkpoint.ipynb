```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import tqdm

def purity_score(y_true, y_pred):
    df = pd.DataFrame({'cluster': y_pred, 'label': y_true})
    correct = ( df
                .groupby('cluster')['label']
                .apply(lambda x: x.value_counts().max())
                .sum() )
    return correct / len(y_true)

def plot_purity_silhouette_vs_ipc_seeds(
    X_original: np.ndarray,
    y:           np.ndarray,
    ipc_values:  list[int] = list(range(10, 101, 10)),
    seeds:       list[int] = [5,6,7,8,9],
    flat = False
):
    # guardamos resultados: dims [n_seeds, n_ipc]
    pur_orig = np.zeros((len(seeds), len(ipc_values)))
    sil_orig = np.zeros_like(pur_orig)
    pur_lat  = np.zeros_like(pur_orig)
    sil_lat  = np.zeros_like(pur_orig)

    for i, seed in enumerate(seeds):
        Z_latent = np.load(f'/content/drive/MyDrive/TFM/Analisis_CorVAE/checkpoint_default_Dlatent_G/vae/train_z_seed_{seed}.npy')
        if flat:
            n, t, d = Z_latent.shape
            Z_latent = Z_latent.reshape(n, t * d)
        else:
            Z_latent = Z_latent[:,0,:]
        for j, ipc in enumerate(tqdm.tqdm(ipc_values)):
            k = 2 * ipc

            # original
            km_o = KMeans(n_clusters=k, random_state=seed, n_init='auto') \
                   .fit(X_original)
            labels_o = km_o.labels_
            pur_orig[i, j] = purity_score(y, labels_o)
            sil_orig[i, j] = silhouette_score(X_original, labels_o)

            # latent
            km_l = KMeans(n_clusters=k, random_state=seed, n_init='auto') \
                   .fit(Z_latent)
            labels_l = km_l.labels_
            pur_lat[i, j] = purity_score(y, labels_l)
            sil_lat[i, j] = silhouette_score(Z_latent, labels_l)

    # calculamos media y std sobre semillas
    mean_pur_orig = pur_orig.mean(axis=0)
    std_pur_orig  = pur_orig.std(axis=0)
    mean_pur_lat  = pur_lat.mean(axis=0)
    std_pur_lat   = pur_lat.std(axis=0)

    mean_sil_orig = sil_orig.mean(axis=0)
    std_sil_orig  = sil_orig.std(axis=0)
    mean_sil_lat  = sil_lat.mean(axis=0)
    std_sil_lat   = sil_lat.std(axis=0)

    # --- Plotting ---
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), sharex=True)

    # Purity
    ax1.plot(ipc_values, mean_pur_orig, marker='o', label='Original')
    ax1.fill_between(ipc_values,
                     mean_pur_orig - std_pur_orig,
                     mean_pur_orig + std_pur_orig,
                     alpha=0.2)
    ax1.plot(ipc_values, mean_pur_lat, marker='o', label='Latent')
    ax1.fill_between(ipc_values,
                     mean_pur_lat - std_pur_lat,
                     mean_pur_lat + std_pur_lat,
                     alpha=0.2)
    ax1.set_title('Purity vs IPC')
    ax1.set_xlabel('Instances Per Class (IPC)')
    ax1.set_ylabel('Purity')
    ax1.legend()
    ax1.grid(True, linestyle='--', alpha=0.5)

    # Silhouette
    ax2.plot(ipc_values, mean_sil_orig, marker='o', label='Original')
    ax2.fill_between(ipc_values,
                     mean_sil_orig - std_sil_orig,
                     mean_sil_orig + std_sil_orig,
                     alpha=0.2)
    ax2.plot(ipc_values, mean_sil_lat, marker='o', label='Latent')
    ax2.fill_between(ipc_values,
                     mean_sil_lat - std_sil_lat,
                     mean_sil_lat + std_sil_lat,
                     alpha=0.2)
    ax2.set_title('Silhouette vs IPC')
    ax2.set_xlabel('Instances Per Class (IPC)')
    ax2.set_ylabel('Silhouette Score')
    ax2.legend()
    ax2.grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
from sklearn.cluster import KMeans

def distill_with_kmeans_cls(
    data: np.ndarray,
    labels: np.ndarray,
    num_centroids: int,
    seed: int = 0
) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Aplica KMeans por clase usando solo el token [CLS] (posición 0) de cada muestra.
    Retorna muestras originales más cercanas a cada centroide.

    Parámetros:
    - data: array de forma [n, f] o [n, t, d]
    - labels: array de etiquetas de longitud n
    - num_centroids: número de centroides por clase
    - seed: semilla para reproducibilidad

    Retorna:
    - final_data: array con las muestras seleccionadas, forma [n_classes * num_centroids, f] o [n_classes * num_centroids, t, d]
    - final_labels: etiquetas correspondientes a cada centroid (longitud n_classes * num_centroids)
    - selected_indices: índices de las muestras originales seleccionadas
    """
    # Extraer solo el token CLS
    if data.ndim == 3:
        n, t, d = data.shape
        cls_tokens = data[:, 0, :]  # [n, d]
    elif data.ndim == 2:
        n, f = data.shape
        cls_tokens = data       # [n, f]
    else:
        raise ValueError(f"distill_with_kmeans_cls: data debe ser 2D o 3D, no {data.ndim}D.")

    selected_points = []
    new_labels = []
    selected_indices = []

    # Clusterizar por clase en espacio CLS
    for class_id in np.unique(labels):
        mask = (labels == class_id)
        class_cls = cls_tokens[mask]            # [n_c, dim]
        original_idxs = np.nonzero(mask)[0]

        kmeans = KMeans(
            n_clusters=num_centroids,
            random_state=seed,
            n_init="auto"
        ).fit(class_cls)

        centers = kmeans.cluster_centers_      # [num_centroids, dim]

        # Calcular distancia L2 CLS->centroides
        dist = np.linalg.norm(
            class_cls[:, None, :] - centers[None, :, :],
            axis=2
        )  # [n_c, num_centroids]
        nearest_pos = dist.argmin(axis=0)     # índice por centroide
        sel_idxs = original_idxs[nearest_pos] # índices en data original

        # Guardar resultados
        selected_indices.extend(sel_idxs.tolist())
        selected_points.append(data[sel_idxs])  # muestras completas
        new_labels.extend([class_id] * num_centroids)

    final_data = np.concatenate(selected_points, axis=0)
    final_labels = np.array(new_labels)
    selected_indices = np.array(selected_indices)

    return final_data, final_labels#, selected_indices
```

```python
def distill_with_agglomerative_cls(
    data: np.ndarray,
    labels: np.ndarray,
    num_clusters: int,
    seed: int = 0
) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Aplica AgglomerativeClustering por clase usando solo el token [CLS] (posición 0).
    Retorna las muestras originales más cercanas a cada cluster.

    Parámetros:
    - data: array de forma [n, f] o [n, t, d]
    - labels: array de etiquetas de longitud n
    - num_clusters: número de clusters por clase
    - seed: semilla para reproducibilidad (no usada por defecto en Agglomerative)

    Retorna:
    - final_data: muestras seleccionadas, forma [n_classes * num_clusters, f] o [n_classes * num_clusters, t, d]
    - final_labels: etiquetas correspondientes
    - selected_indices: índices de las muestras originales seleccionadas
    """
    # Extraer solo el token CLS
    if data.ndim == 3:
        n, t, d = data.shape
        cls_tokens = data[:, 0, :]  # [n, d]
    elif data.ndim == 2:
        n, f = data.shape
        cls_tokens = data       # [n, f]
    else:
        raise ValueError(f"distill_with_agglomerative_cls: data debe ser 2D o 3D, no {data.ndim}D.")

    selected_points = []
    new_labels = []
    selected_indices = []

    # Clusterizar por clase en espacio CLS
    for class_id in np.unique(labels):
        mask = (labels == class_id)
        class_cls = cls_tokens[mask]            # [n_c, dim]
        orig_idxs = np.nonzero(mask)[0]

        clustering = AgglomerativeClustering(
            n_clusters=num_clusters
        ).fit(class_cls)

        # Calcular centroides como media de cada cluster
        centers = []
        for k in range(num_clusters):
            pts = class_cls[clustering.labels_ == k]
            centers.append(pts.mean(axis=0))
        centers = np.stack(centers, axis=0)      # [num_clusters, dim]

        # Para cada centroide, elegimos la muestra CLS más cercana
        dist = np.linalg.norm(
            class_cls[:, None, :] - centers[None, :, :],
            axis=2
        )  # [n_c, num_clusters]
        nearest_pos = dist.argmin(axis=0)       # índice en class_cls por cluster
        sel_idxs = orig_idxs[nearest_pos]       # índices en data original

        # Guardar resultados
        selected_indices.extend(sel_idxs.tolist())
        selected_points.append(data[sel_idxs])  # muestras completas
        new_labels.extend([class_id] * num_clusters)

    final_data = np.concatenate(selected_points, axis=0)
    final_labels = np.array(new_labels)
    selected_indices = np.array(selected_indices)

    return final_data, final_labels#, selected_indices
```
