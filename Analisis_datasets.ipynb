# Análisis del dataset shoppers

Contiene un analisis a nivel de distribución tanto de columnas numéricas como categóricas, así como
también sobre matrices de correlación, valores faltantes, desbalance de clases y valores atípicos

```python
df_shoppers = pd.read_csv('data/shoppers/dataset.csv')

# 3. Vista inicial
# 3.1. Mostrar las primeras filas
print("Primeras 5 filas:")
display(df_shoppers.head())

# 3.2. Información general de columnas, tipos y valores no nulos
print("\nInformación del DataFrame:")
display(df_shoppers.info())

# 3.3. Estadísticos descriptivos de variables numéricas
print("\nEstadísticos descriptivos (numéricos):")
display(df_shoppers.describe().T)

# 3.4. Estadísticos descriptivos de variables categóricas
print("\nEstadísticos descriptivos (categóricos):")
display(df_shoppers.describe(include=['object', 'category']).T)

# 4. Detección de valores faltantes
print("\nValores faltantes por columna:")
display(df_shoppers.isnull().sum())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar tema y contexto para publicaciones en paper
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)

# Cargar el dataset
df_shoppers = pd.read_csv('data/shoppers/dataset.csv')

# Identificar variables numéricas y categóricas
num_cols = df_shoppers.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_shoppers.select_dtypes(include=['object', 'category']).columns.tolist()

# 1. Histogramas + KDE para variables numéricas
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df_shoppers[col].dropna(), kde=True, color="steelblue", edgecolor="black")
    plt.title(f"Distribución de {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frecuencia", fontsize=12)
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()
```

```python


# 2. Countplot para variables categóricas
for col in cat_cols:
    plt.figure(figsize=(6, 4))
    order = df_shoppers[col].value_counts().index
    sns.countplot(data=df_shoppers, x=col, order=order, palette="pastel", edgecolor="black")
    plt.title(f"Conteo de {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frecuencia", fontsize=12)
    plt.xticks(rotation=45, ha='right')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

```

```python
# Identificar variables numéricas y categóricas
num_cols = df_shoppers.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_shoppers.select_dtypes(include=['object', 'category']).columns.tolist()


# 3. Mapa de calor de correlaciones entre variables numéricas
plt.figure(figsize=(20, 8))
corr = df_shoppers[num_cols].corr()
sns.heatmap(corr, annot=True, fmt=".1f", cmap="vlag", cbar=True, square=True, linewidths=.5)
plt.title("Mapa de calor de shopper", fontsize=16)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.savefig('results/exploratory_analysis/correlaciones_shopper.png')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar tema y contexto para publicaciones en paper
#sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)

# Identificar variables numéricas y categóricas
num_cols = df_shoppers.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_shoppers.select_dtypes(include=['object', 'category']).columns.tolist()

# 1. Histogramas + KDE para variables numéricas
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df_shoppers[col].dropna(), kde=True, color="steelblue", edgecolor="black")
    plt.title(f"Distribución de {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frecuencia", fontsize=12)
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

# 2. Countplot para variables categóricas
for col in cat_cols:
    plt.figure(figsize=(6, 4))
    order = df_shoppers[col].value_counts().index
    sns.countplot(data=df_shoppers, x=col, order=order, palette="pastel", edgecolor="black")
    plt.title(f"Conteo de {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Frecuencia", fontsize=12)
    plt.xticks(rotation=45, ha='right')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

# 4. Boxplots para detectar outliers en variables numéricas
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df_shoppers[col], color="lightseagreen")
    plt.title(f"Boxplot de {col}", fontsize=14)
    plt.xlabel(col, fontsize=12)
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

# 5. Ejemplo de boxplot bivariado (numérico vs. categórico)
if len(cat_cols) > 0 and len(num_cols) > 0:
    ejemplo_cat = cat_cols[0]
    ejemplo_num = num_cols[0]
    plt.figure(figsize=(8, 5))
    sns.boxplot(x=ejemplo_cat, y=ejemplo_num, data=df_shoppers, palette="Set2")
    plt.title(f"{ejemplo_num} por {ejemplo_cat}", fontsize=16)
    plt.xlabel(ejemplo_cat, fontsize=12)
    plt.ylabel(ejemplo_num, fontsize=12)
    plt.xticks(rotation=45, ha='right')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Identificar variables numéricas
num_cols = df_shoppers.select_dtypes(include=['number']).columns.tolist()
n = len(num_cols)

# Definir disposición de subplots: hasta 3 columnas por fila, ajustar filas
max_cols_per_row = 3
n_cols_subplot = min(n, max_cols_per_row)
n_rows_subplot = math.ceil(n / n_cols_subplot)

# Crear figura con subplots
fig, axes = plt.subplots(
    n_rows_subplot,
    n_cols_subplot,
    figsize=(4 * n_cols_subplot, 4 * n_rows_subplot),
    sharey=False
)

# Aplanar la matriz de ejes para iterar fácilmente
axes_flat = axes.flatten() if n > 1 else [axes]

# Generar cada boxplot en su eje correspondiente
for idx, col in enumerate(num_cols):
    ax = axes_flat[idx]
    sns.boxplot(
        x=df_shoppers[col],
        ax=ax,
        color="lightseagreen"
    )
    ax.set_title(f"{col}", fontsize=12)
    ax.set_xlabel("")  # Quitamos etiqueta del eje x (solo aparece el nombre en el título)
    ax.set_ylabel("")  # Opcional: si prefieres solo números sin etiqueta

# Desactivar subplots sobrantes (si n no es múltiplo de 3)
for j in range(n, len(axes_flat)):
    axes_flat[j].axis('off')

# --------------------------------------------------------------------------------
# Importante: fijar el título principal *antes* de llamar a tight_layout, 
# y decirle a tight_layout que reserve espacio para el suptitle vía `rect`.
# --------------------------------------------------------------------------------
plt.suptitle("Boxplots de Variables Numéricas", fontsize=16)

# Ajustamos tight_layout dejando espacio superior para el suptitle.
# El rect=[0, 0, 1, 0.95] indica que el área de subplots ocupe desde 
# el 0% al 95% de la altura de la figura (dejando el 5% arriba para el título).
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Guardar figura asegurándonos de que todo entre en los límites
plt.savefig('results/exploratory_analysis/boxplot_adult.png', bbox_inches='tight')

# Finalmente mostramos en pantalla
plt.show()
```

```python
# --------------------------------------------------
# 1. Cargar los tres datasets y definir su target
# --------------------------------------------------

# 1.1. Shoppers (ya cargado antes); target: 'income'
df_shoppers = pd.read_csv('data/shoppers/dataset.csv')
target_shoppers = 'Revenue'

# 1.2. Adult; target: 'income'
df_adult = pd.read_csv('data/adult/dataset.csv')
df_adult['income'] = df_adult['income'].str.replace('.', '', regex=False)
target_adult = 'income'

# 1.3. Default; target: 'default'
df_default = pd.read_csv('data/default/dataset.csv')
target_default = 'Y'

# --------------------------------------------------
# 2. Preparar figura y ejes (1 fila, 3 columnas)
# --------------------------------------------------

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(18, 5),
    sharey=False
)

# --------------------------------------------------
# 3. Plot #1: Shoppers – Conteo de 'income'
# --------------------------------------------------

ax = axes[0]
order_sh = df_shoppers[target_shoppers].value_counts().index
sns.countplot(
    data=df_shoppers,
    x=target_shoppers,
    order=order_sh,
    palette="pastel",
    edgecolor="black",
    ax=ax
)

ax.set_title("Conteo de Shoppers", fontsize=14)
ax.set_xlabel("Revenue", fontsize=12)
ax.set_ylabel("Frecuencia", fontsize=12)
ax.tick_params(axis='x', rotation=45)
sns.despine(ax=ax, trim=True)

# --------------------------------------------------
# 4. Plot #2: Adult – Conteo de 'income'
# --------------------------------------------------

ax = axes[1]
order_ad = df_adult[target_adult].value_counts().index
sns.countplot(
    data=df_adult,
    x=target_adult,
    order=order_ad,
    palette="pastel",
    edgecolor="black",
    ax=ax
)

ax.set_title("Conteo de Adult", fontsize=14)
ax.set_xlabel("Income", fontsize=12)
ax.set_ylabel("")  # Ya no necesitamos repetir etiqueta y-error
ax.tick_params(axis='x', rotation=45)
sns.despine(ax=ax, trim=True)

# --------------------------------------------------
# 5. Plot #3: Default – Conteo de 'default'
# --------------------------------------------------

ax = axes[2]
order_df = df_default[target_default].value_counts().index
sns.countplot(
    data=df_default,
    x=target_default,
    order=order_df,
    palette="pastel",
    edgecolor="black",
    ax=ax
)

ax.set_title("Conteo de Default", fontsize=14)
ax.set_xlabel("Y", fontsize=12)
ax.set_ylabel("")  # Mismas razones que en el subplot anterior
ax.tick_params(axis='x', rotation=45)
sns.despine(ax=ax, trim=True)

# --------------------------------------------------
# 6. Ajustes finales y guardado de la figura
# --------------------------------------------------

plt.tight_layout(rect=[0, 0, 1, 0.95])  # deja espacio para un posible suptitle
# Si quieres un título general descomenta la línea siguiente:
# plt.suptitle("Distribución de las variables objetivo en tres datasets", fontsize=16, y=1.02)
plt.savefig('results/exploratory_analysis/target_dis.png')
plt.show()
```
